# 📊 Histogram

Histogram udáva, koľko pixelov má jednotlivé hodnoty intenzity:

- vodorovná os: intenzita od 0 do 255,
- zvislá os: počet pixelov,
- prevaha hodnôt vľavo zodpovedá skôr tmavému obrazu,
- prevaha hodnôt vpravo skôr svetlému obrazu,
- úzky rozsah hodnôt často signalizuje nízky kontrast.

Funkcia `cv.calcHist([gray], [0], None, [256], [0, 256])` vytvorí 256 košov pre hodnoty jedného kanála. Parameter `None` znamená, že používame celý obraz bez masky.

````python
def show_image_and_histogram(image, title="Obraz"):
    hist = cv.calcHist([image], [0], None, [256], [0, 256]).ravel()
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))
    axes[0].imshow(image, cmap="gray", vmin=0, vmax=255)
    axes[0].set_title(title)
    axes[0].axis("off")

    axes[1].plot(np.arange(256), hist, color="black")
    axes[1].set_title("Histogram intenzít")
    axes[1].set_xlabel("Intenzita")
    axes[1].set_ylabel("Počet pixelov")
    axes[1].set_xlim(0, 255)
    axes[1].grid(alpha=0.2)
    plt.tight_layout()
    plt.show()

    return hist
````

````python
dark = np.clip(gray.astype(np.float32) * 0.45, 0, 255).astype(np.uint8)
bright = np.clip(gray.astype(np.float32) * 0.65 + 90, 0, 255).astype(np.uint8)
low_contrast = np.clip((gray.astype(np.float32) - 127) * 0.30 + 127, 0, 255).astype(np.uint8)

variants = [(dark, "Tmavý"), (bright, "Svetlý"), (low_contrast, "Nízky kontrast")]
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for column, (image, title) in enumerate(variants):
    axes[0, column].imshow(image, cmap="gray", vmin=0, vmax=255)
    axes[0, column].set_title(title)
    axes[0, column].axis("off")
    hist = cv.calcHist([image], [0], None, [256], [0, 256])
    axes[1, column].plot(hist, color="black")
    axes[1, column].set_xlim(0, 255)
    axes[1, column].set_xlabel("Intenzita")
    axes[1, column].set_ylabel("Počet pixelov")
    axes[1, column].grid(alpha=0.2)
plt.tight_layout()
plt.show()
````

### Ekvalizácia (vyhladzovanie) histogramu

`cv.equalizeHist()` transformuje intenzity tak, aby lepšie využili dostupný rozsah. Môže zvýšiť globálny kontrast, najmä pri obraze s hodnotami sústredenými v úzkom intervale.

Ekvalizácia ale nie je univerzálne „vylepšenie“. Môže zvýrazniť šum, vytvoriť neprirodzený vzhľad alebo zhoršiť lokálne detaily. Výsledok preto vždy hodnotíme vizuálne s ohľadom na ďalšiu úlohu spracovania.

````python
equalized = cv.equalizeHist(gray)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for column, (image, title) in enumerate([
    (gray, "Pred ekvalizáciou"),
    (equalized, "Po globálnej ekvalizácii"),
]):
    axes[0, column].imshow(image, cmap="gray", vmin=0, vmax=255)
    axes[0, column].set_title(title)
    axes[0, column].axis("off")
    hist = cv.calcHist([image], [0], None, [256], [0, 256])
    axes[1, column].plot(hist, color="black")
    axes[1, column].set_xlim(0, 255)
    axes[1, column].set_xlabel("Intenzita")
    axes[1, column].set_ylabel("Počet pixelov")
    axes[1, column].grid(alpha=0.2)
plt.tight_layout()
plt.show()
````

CLAHE (*Contrast Limited Adaptive Histogram Equalization*) rozdelí obraz na menšie oblasti a upraví ich lokálne. Parameter `clipLimit` obmedzuje prílišné zosilnenie kontrastu a šumu. `tileGridSize` určuje počet lokálnych oblastí.

````python
clahe = cv.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_result = clahe.apply(gray)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
for ax, image, title in zip(
    axes,
    [gray, equalized, clahe_result],
    ["Originál", "Globálna ekvalizácia", "CLAHE"],
):
    ax.imshow(image, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")
    
plt.tight_layout()
plt.show()
````
